# Black Scholes Option Pricing Model

## Introduction
The Black-Scholes model is a mathematical model for pricing European-style options. It was developed by Fischer Black, Myron Scholes, and Robert Merton in the early 1970s. The model provides a theoretical estimate of the price of an option based on several factors:

- $S$: Current price of the underlying asset
- $K$: Strike price of the option
- $T$: Time to maturity (in years)
- $r$: Risk-free interest rate (annualized)
- $\sigma$: Volatility of the underlying asset (annualized)

The formula for the price of a European call option is given by:
$$C = S \cdot N(d_1) - K \cdot e^{-rT} \cdot N(d_2)$$

And for a European put option, the price is given by:
$$P = K \cdot e^{-rT} \cdot N(-d_2) - S \cdot N(-d_1)$$

Where:
$$d_1 = \frac{\ln(S/K) + (r + \sigma^2/2)T}{\sigma \sqrt{T}}$$
$$d_2 = d_1 - \sigma \sqrt{T}$$

> Note: $N(x)$ is the cumulative distribution function of the standard normal distribution.
